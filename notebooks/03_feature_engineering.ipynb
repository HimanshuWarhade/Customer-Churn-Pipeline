```python
import pandas as pd

df = pd.read_csv("../data/raw/Customer-Churn-Records.csv")
df.shape
```

```python
df = df.drop(columns=["RowNumber", "CustomerId", "Surname"])
df.columns.tolist()
```

```python
df = df.drop(columns=["Complain"])
df.shape
```

```python
df["HasZeroBalance"] = (df["Balance"] == 0).astype(int)
df["HasZeroBalance"].value_counts()
```

```python
df["BalanceSalaryRatio"] = df["Balance"] / (df["EstimatedSalary"] + 1)
df[["Balance", "EstimatedSalary", "BalanceSalaryRatio"]].head()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 30, 45, 60, 100],
    labels=["18-30", "31-45", "46-60", "60+"],
)
df["AgeGroup"].value_counts()
```

```python
categorical_cols = ["Geography", "Gender", "Card Type", "AgeGroup"]

df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
df.shape
```

```python
print("Final shape:", df.shape)
print()
print("Column dtypes:")
print(df.dtypes)
```

```python
df.to_csv("../data/processed/featured_data.csv", index=False)
print("Saved to data/processed/featured_data.csv")
```
